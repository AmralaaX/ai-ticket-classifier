```python
# Step 1: Create the function
def classify_ticket(ticket):

    # Convert the message to lowercase
    message = ticket["message"].lower()

    # Check the category
    if "delivery" in message:
        ticket["category"] = "Delivery"

    elif "payment" in message:
        ticket["category"] = "Payment"

    else:
        ticket["category"] = "Other"

    # Check the priority
    if ticket["amount"] >= 5000:
        ticket["priority"] = "High"

    else:
        ticket["priority"] = "Normal"

    # Return the classified ticket
    return ticket


# Step 2: Create a list of 3 tickets
tickets = [
    {
        "customer": "Sara",
        "message": "My payment failed",
        "amount": 7500
    },
    {
        "customer": "Omar",
        "message": "My delivery is late",
        "amount": 1500
    },
    {
        "customer": "Ali",
        "message": "I cannot login",
        "amount": 0
    }
]


# Step 3: Loop through all tickets
for ticket in tickets:

    # Send each ticket to the function
    result = classify_ticket(ticket)

    # Print the classified ticket
    print(result)
```
